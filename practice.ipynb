## Part 2 — What the columns are stored as

````python
sales_2009.dtypes
````

Every column has one type, and the type decides what you can do with the values in it.
`Quantity` is whole numbers, so you can add them up. `InvoiceDate` came in as text, so at
the moment you can only compare it letter by letter. Nothing in it knows about months.

A **timestamp** is the type that does know. One timestamp is a single moment: the year, the
month, the day, the hour, the minute and the second of one event, all in one value.
`pd.to_datetime` reads text and gives you back timestamps.

Some methods only make sense for one type, so pandas keeps them behind a door you have to
open by name:

- `.dt` opens the timestamp methods, on a column of timestamps.
- `.str` opens the text methods, on a column of text.

Ask for one on the wrong type and pandas refuses.

The same split exists for lists, and you can try it on a short piece of text:

### First copies and later copies

`duplicated()` goes down the table and marks a row `True` if it has seen that exact row
before. The first row of a group is therefore marked `False`.

Six rows make it concrete. Two of them are the same, three others are the same, one is

| | Invoice | Item | `duplicated()` |

| 0 | A | x | False |
| 1 | A | x | **True** |
| 2 | B | y | False |

| 4 | C | z | **True** |
| 5 | C | z | **True** |

### A code column that holds more than codes

Nearly every `StockCode` begins with digits. A lot of them then carry a letter, like
`85123A`, which is one product in another size or colour. A few begin with a letter
instead.

`str[0]` takes the first character of every value in a column, and `str.isdigit()` asks of
every value whether it is made of digits. They chain, and **each one needs its own `.str`**:

```
sales["StockCode"].str[0].str.isdigit()
```

single piece of text and you are asking a whole column.
